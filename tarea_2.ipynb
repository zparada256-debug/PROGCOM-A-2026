```python
#16
num = int(input("Introduce el número para ver su tabla: "))
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

```python
#17
n = int(input("Sumar hasta el número: "))
suma = 0
for i in range(1, n + 1):
    suma += i
print(f"La suma total es: {suma}")
```

```python
#18
num = int(input("Introduce un número para el factorial: "))
if num < 0:
    print("No existen factoriales de números negativos.")
else:
    factorial = 1
    for i in range(1, num + 1):
        factorial *= i
    print(f"El factorial de {num} es {factorial}")
```

```python
#19
n = int(input("¿Cuántos números de Fibonacci quieres ver? "))
a, b = 0, 1
for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

```python
#20
colores = []
print("Define 10 colores:")
for i in range(10):
    colores.append(input(f"Color {i+1}: ").lower())

secreto = colores[0] # El programa elige el primero como el secreto
print("\n--- ¡Adivina el color secreto! ---")

for intento in range(1, 11):
    pista = input(f"Intento {intento}: ").lower()
    if pista == secreto:
        print("¡Felicidades! Adivinaste.")
        break
    else:
        print("Ese no es.")
```

```python
#23
num = int(input("Introduce un número: "))
es_primo = True
if num < 2:
    es_primo = False
else:
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            es_primo = False
            break

if es_primo:
    print(f"{num} es primo.")
else:
    print(f"{num} no es primo.")
```
